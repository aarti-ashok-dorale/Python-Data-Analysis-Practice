## Type casting = convert one data type into another

# 1. implicit 
# 2. explicit

### 1.implicit = no need to convert datatypes, python do it itself

```python
a = 12
b = 1.2
print(type(a))
print(type(b))

print(a+b)
```

### 2.Explicit  = user can change datatype whatever they want.

```python
a = "20"
b = 25
a = int(a)

c = a+b
print(c)
```

#### swap two variables

```python
a = 10
b = 20
temp = a
a = b
b = temp

print(a,b)

```

```python
#### swap 3 variables

a = 10
b = 20
c = 30
temp = a
a = b
b = c
c = temp

print(a,b,c)
```
