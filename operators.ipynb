##### 1. Arithematic 

```python
a = 10
b = 30

add = a+b
sub = a-b
multiplication = a*b
divide = a/b
power = a**b
floor_division = a//b
modulus = a%b

print(add, sub, multiplication, divide, power, floor_division, modulus)
```

##### 2. comparison or relational 

```python
x = 25
y = 56

less_than = x<y
greater_than = x>y
equal_to = x==y
lessthan_equalto = x<=y
greaterthan_equalto = x>=y
not_equalto = x!=y

print(less_than, greater_than, equal_to, lessthan_equalto, greaterthan_equalto, not_equalto)
```

##### 3. logical = only give true and false  output and combine multiple conditions

```python
# and 
print(3>7 and 4<2)
print(6>0 and 5>2)
```

```python
#or
print(3==3 or 6>9)
print(34>67 or 98>100)
```

```python
#not

print(not(9<2 and 0<2))
print(not(8>2 and 1<2))
```

##### 4. Assignment = it will assign and update the value of variable

```python
a = 5
b = 10

a = b
print(a)      

a += b
print(a)    

a -= b
print(a)     

a *= b
print(a)    

a /= b
print(a)    

a //= b
print(a)    

a **= b
print(a)    

a %= b
print(a)     
```
